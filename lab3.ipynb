#### Проведите все три волны отбора студентов на курсы по выбору Результат сохраните в файл res_fall.csv.

```python
result = df[['id', 'percentile', 'fall_course_num', 'fall_1', 'fall_2', 'fall_3']].copy()
result['course1'] = None
result['course2'] = None

courses = pd.unique(result[['fall_1', 'fall_2', 'fall_3']].values.ravel())
seats = pd.Series(30.0, index=courses)
seats['Statistical Learning Theory'] = 60
seats['Высокопроизводительные вычисления'] = 60
seats['Анализ неструктурированных данных'] = np.inf

two = result['fall_course_num'] == 2
wave_fall_1 = 1
wave_fall_2 = np.where(two, 1, 2)
wave_fall_3 = np.where(two, 2, 3)
```

```python
for wave in [1, 2, 3]:
    for course in courses:
        got = result['course1'].notna().astype(int) + result['course2'].notna().astype(int)
        wants = (((result['fall_1'] == course) & (wave_fall_1 == wave)) |
                 ((result['fall_2'] == course) & (wave_fall_2 == wave)) |
                 ((result['fall_3'] == course) & (wave_fall_3 == wave)))
        candidates = result[wants & (got < result['fall_course_num'])]

        taken = (result['course1'] == course).sum() + (result['course2'] == course).sum()
        n_free = int(min(seats[course] - taken, len(candidates)))
        winners = candidates.sort_values('percentile').index[:n_free]

        has_first = result.loc[winners, 'course1'].notna()
        result.loc[winners[~has_first], 'course1'] = course
        result.loc[winners[has_first], 'course2'] = course
```

```python
result['is_first_place'] = False
result['is_last_place'] = False
for course in courses:
    in_course = (result['course1'] == course) | (result['course2'] == course)
    pct = result.loc[in_course, 'percentile']
    result.loc[in_course & (result['percentile'] == pct.min()), 'is_first_place'] = True
    result.loc[in_course & (result['percentile'] == pct.max()), 'is_last_place'] = True

result['is_first_place'] = result['is_first_place'].map({True: True, False: np.nan})
result['is_last_place'] = result['is_last_place'].map({True: True, False: np.nan})

res_fall = pd.DataFrame({
    'ID': result['id'],
    'course1': result['course1'].fillna('???'),
    'course2': np.where(result['fall_course_num'] == 1, '-', result['course2'].fillna('???'))
})
res_fall.to_csv('res_fall.csv', index=None)
```
